```python
text_columns = ["Item", "Payment Method", "Location"]

for col in text_columns:
    df[col] = df[col].str.strip()
```

```python
print(df["Item"].unique())
print(df["Payment Method"].unique())
print(df["Location"].unique())
```

```python
cols = ["Item", "Payment Method", "Location"]

for col in cols:
    df[col] = df[col].replace(["UNKNOWN", "ERROR"], pd.NA)
```

```python
df[["Item", "Payment Method", "Location"]].isnull().sum()
```

```python
print(df["Item"].unique())
print(df["Payment Method"].unique())
print(df["Location"].unique())
```

```python
df[["Quantity", "Price Per Unit", "Total Spent"]].describe()
```

```python
numeric_columns = ["Quantity", "Price Per Unit", "Total Spent"]

for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")
```

```python
df[["Quantity", "Price Per Unit", "Total Spent"]].dtypes
```

```python
df[["Quantity", "Price Per Unit", "Total Spent"]].describe()
```

```python
df["Calculated Total"] = df["Quantity"] * df["Price Per Unit"]

df[["Quantity", "Price Per Unit", "Total Spent", "Calculated Total"]].head(10)
```

```python
comparison = df["Total Spent"] == df["Calculated Total"]

comparison.value_counts(dropna=False)
```

```python
mismatch = df[
    df["Total Spent"].notna() &
    df["Calculated Total"].notna() &
    (df["Total Spent"] != df["Calculated Total"])
]

print("Mismatched rows:", len(mismatch))
```

```python
missing_total = df[
    df["Total Spent"].isna() &
    df["Quantity"].notna() &
    df["Price Per Unit"].notna()
]

print("Missing Total Spent with enough data:", len(missing_total))
```

```python
df["Total Spent"] = df["Total Spent"].fillna(
    df["Quantity"] * df["Price Per Unit"]
)
```

```python
df["Total Spent"].isna().sum()
```

```python
df[["Quantity", "Price Per Unit"]].isnull().sum()
```

```python
print(
    "Quantity recoverable:",
    df["Quantity"].isna().sum()
    - (df["Quantity"].isna() & df["Total Spent"].isna()).sum()
)

print(
    "Price Per Unit recoverable:",
    df["Price Per Unit"].isna().sum()
    - (df["Price Per Unit"].isna() & df["Total Spent"].isna()).sum()
)
```

```python
df["Calculated Quantity"] = (
    df["Total Spent"] / df["Price Per Unit"]
)

df["Calculated Price"] = (
    df["Total Spent"] / df["Quantity"]
)
```

```python
print(df[[
    "Quantity",
    "Price Per Unit",
    "Total Spent",
    "Calculated Quantity",
    "Calculated Price"
]].isnull().sum())
```

```python
print(df.loc[
    df["Quantity"].isna(),
    ["Quantity", "Price Per Unit", "Total Spent", "Calculated Quantity", "Calculated Price"]
].head(10))
```

```python
valid_quantity = df.loc[
    df["Quantity"].isna() & df["Calculated Quantity"].notna(),
    "Calculated Quantity"
]

print("Recoverable:", len(valid_quantity))
print("Non-integer:", (valid_quantity % 1 != 0).sum())
print("Below 1:", (valid_quantity < 1).sum())
print("Above 5:", (valid_quantity > 5).sum())
```

```python
df["Quantity"] = df["Quantity"].fillna(
    df["Calculated Quantity"]
)
```

```python
print(df["Quantity"].isna().sum())
```

```python
valid_price = df.loc[
    df["Price Per Unit"].isna() & df["Calculated Price"].notna(),
    "Calculated Price"
]

print("Recoverable:", len(valid_price))
print("Non-integer:", (valid_price % 1 != 0).sum())
print("Below 1:", (valid_price < 1).sum())
print("Above 5:", (valid_price > 5).sum())
```

```python
non_integer_prices = df.loc[
    df["Price Per Unit"].isna() &
    df["Calculated Price"].notna() &
    (df["Calculated Price"] % 1 != 0),
    ["Quantity", "Total Spent", "Calculated Price"]
]

print(non_integer_prices.head(20))
```

```python
df["Price Per Unit"] = df["Price Per Unit"].fillna(
    df["Calculated Price"]
)
```

```python
print(df["Price Per Unit"].isna().sum())
```

```python
remaining_missing = df.loc[
    df["Quantity"].isna() |
    df["Price Per Unit"].isna() |
    df["Total Spent"].isna(),
    [
        "Transaction ID",
        "Quantity",
        "Price Per Unit",
        "Total Spent"
    ]
]

print(remaining_missing.head(20))
print("\nRows remaining:", len(remaining_missing))
```

```python
print(
    df.loc[
        df["Quantity"].isna() &
        df["Price Per Unit"].isna() &
        df["Total Spent"].notna()
    ].shape[0],
    "rows: Total only"
)

print(
    df.loc[
        df["Quantity"].notna() &
        df["Price Per Unit"].isna() &
        df["Total Spent"].isna()
    ].shape[0],
    "rows: Quantity only"
)

print(
    df.loc[
        df["Quantity"].isna() &
        df["Price Per Unit"].notna() &
        df["Total Spent"].isna()
    ].shape[0],
    "rows: Price only"
)

print(
    df.loc[
        df["Quantity"].isna() &
        df["Price Per Unit"].isna() &
        df["Total Spent"].isna()
    ].shape[0],
    "rows: All three missing"
)
```

```python
print("UNKNOWN:", (df["Transaction Date"] == "UNKNOWN").sum())
print("ERROR:", (df["Transaction Date"] == "ERROR").sum())
print("NaT:", df["Transaction Date"].isna().sum())
```

```python
df["Transaction Date"] = df["Transaction Date"].replace(
    ["UNKNOWN", "ERROR"],
    pd.NaT
)
```

```python
df["Transaction Date"] = pd.to_datetime(
    df["Transaction Date"],
    errors="coerce"
)
```

```python
print("Missing dates:", df["Transaction Date"].isna().sum())
print("Minimum date:", df["Transaction Date"].min())
print("Maximum date:", df["Transaction Date"].max())
print("Data type:", df["Transaction Date"].dtype)
```

```python
df = df.drop(
    columns=[
        "Calculated Total",
        "Calculated Quantity",
        "Calculated Price"
    ]
)
```
